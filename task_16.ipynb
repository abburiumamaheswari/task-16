```python
import pandas as pd

# Creating a student DataFrame
data = {
    'Name': ['Uma', 'Ravi', 'Sita', 'Ram', 'Priya'],
    'Age': [19, 20, 18, 21, 19],
    'Branch': ['CSE', 'ECE', 'CSE', 'IT', 'DS'],
    'Marks': [85, 90, 78, 88, 95]
}

df = pd.DataFrame(data, index=['S101', 'S102', 'S103', 'S104', 'S105'])

print("Original DataFrame:")
print(df)
```

```python
# 1. Select a single row
print(df.loc['S101'])

# 2. Select multiple rows
print(df.loc[['S101', 'S103']])

# 3. Select a single column
print(df.loc[:, 'Name'])

# 4. Select specific rows and columns
print(df.loc[['S101', 'S102'], ['Name', 'Marks']])

# 5. Select rows using label slicing
print(df.loc['S101':'S103', ['Name', 'Age']])
```

```python
# 1. Select a single row
print(df.iloc[0])

# 2. Select multiple rows
print(df.iloc[[0, 2]])

# 3. Select a single column
print(df.iloc[:, 0])

# 4. Select specific rows and columns
print(df.iloc[[0, 1], [0, 3]])

# 5. Select rows using position slicing
print(df.iloc[0:3, 0:2])
```

```python
import pandas as pd

data = {
    'Employee': ['Asha', 'Ravi', 'Kiran', 'Priya', 'Arun'],
    'Department': ['HR', 'IT', 'Sales', 'IT', 'HR'],
    'Salary': [25000, 40000, 30000, 45000, 28000]
}

df = pd.DataFrame(data, index=['E1', 'E2', 'E3', 'E4', 'E5'])

print("IT Employees:")
print(df.loc[df['Department'] == 'IT'])

print("Employees with salary above 30000:")
print(df.loc[df['Salary'] > 30000])

print("Employee names and salaries:")
print(df.loc[:, ['Employee', 'Salary']])
```
